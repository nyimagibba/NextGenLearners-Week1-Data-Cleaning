```python

import pandas as pd

# 1. Load the original dataset
df = pd.read_csv('applicants.csv')

print("Original dataset:", df.shape)

# 2. Remove records with missing applicant names
df = df.dropna(subset=['Applicant Name']).copy()

# 3. Handle missing values
df['Email'] = df['Email'].fillna('Not Provided')
df['University'] = df['University'].fillna('Not Specified')
df['Status'] = df['Status'].fillna('Under Review')

# 4. Remove exact duplicates
df = df.drop_duplicates().copy()

# 5. Standardize email addresses
df['Email'] = df['Email'].str.strip().str.lower()

# 6. Standardize internship domains
df['Domain Applied'] = (
    df['Domain Applied'].str.strip().str.lower()
)

domain_mapping = {
    'web dev': 'Web Development',
    'webdev': 'Web Development',
    'web development': 'Web Development',
    'data science': 'Data Science',
    'cybersecurity': 'Cybersecurity',
    'ui/ux design': 'UI/UX Design',
    'mobile app development': 'Mobile App Development'
}

df['Domain Applied'] = df['Domain Applied'].map(domain_mapping)

# 7. Standardize application statuses
df['Status'] = df['Status'].str.strip().str.title()

# 8. Standardize applicant names
df['Applicant Name'] = (
    df['Applicant Name']
    .str.strip()
    .str.replace(r'\s+', ' ', regex=True)
    .str.title()
)

# 9. Standardize university names
df['University'] = (
    df['University']
    .str.strip()
    .str.replace(r'\s+', ' ', regex=True)
)

# 10. Convert application dates
df['Application Date'] = pd.to_datetime(
    df['Application Date'],
    format='mixed',
    errors='coerce'
)

# 11. Standardize phone numbers
df['Phone'] = (
    df['Phone']
    .astype(str)
    .str.replace(r'[\s-]', '', regex=True)
)

# 12. Remove duplicate valid emails
valid_email = (
    df['Email'].notna() &
    df['Email'].ne('not provided')
)

df = pd.concat([
    df.loc[valid_email].drop_duplicates(
        subset=['Email'], keep='first'
    ),
    df.loc[~valid_email]
]).sort_index()

# 13. Reset the index
df = df.reset_index(drop=True)

print("Cleaned dataset:", df.shape)
print("Cleaning completed successfully!")
```

Verify Our Results

```python

print("FINAL DATA QUALITY REPORT")
print("-------------------------")

print("Total records:", len(df))
print("Total columns:", df.shape[1])

print("\nMissing values:")
print(df.isnull().sum())

print("\nExact duplicates:", df.duplicated().sum())
print("Unique domains:", df['Domain Applied'].nunique())
print("Unique statuses:", df['Status'].nunique())
print("Date data type:", df['Application Date'].dtype)

print("Phone formatting issues:",
      df['Phone'].str.contains(
          r'[\s-]', regex=True, na=False
      ).sum())
```
